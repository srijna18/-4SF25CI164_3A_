Pandas

```python
import pandas as pd
import numpy as np
```

1.

```python
df = pd.DataFrame(np.random.randint(1, 10, (5, 3)),columns=['A', 'B', 'C'])
print(df)
df['Product'] = df['A'] * df['B']
print(df)
```

2.

```python
df = pd.DataFrame(np.random.randint(1, 10, (4, 3)))
print(df)
print("Row sum:")
print(df.sum(axis=1))
print("Column sum:")
print(df.sum(axis=0))
```

3.

```python
df = pd.DataFrame(np.random.randint(1, 10, (5, 3)),columns=['A', 'B', 'C'])
df.iloc[1, 0] = np.nan
df.iloc[2, 1] = np.nan
print(df)
df = df.fillna(df.mean())
print(df)
```

4.

```python
df = pd.DataFrame(np.random.randint(1, 10, (6, 4)))
df.iloc[1, 0] = np.nan
df.iloc[3, 2] = np.nan
print(df)
df = df.dropna()
print(df)
```

5.

```python
df = pd.DataFrame({'Category': np.random.choice(['A', 'B', 'C'], 10),'Value': np.random.randint(1, 100, 10)})
print(df)
result = df.groupby('Category')['Value'].agg(['sum', 'mean'])
print(result)
```

6.

```python
df = pd.DataFrame({
    'Product': ['Pen', 'Book', 'Pencil', 'Bag', 'Pen'],
    'Category': ['A', 'B', 'A', 'B', 'A'],
    'Sales': [100, 200, 150, 300, 250]
})
print(df)
result = df.groupby('Category')['Sales'].sum()
print(result)
```

7.

```python
df1 = pd.DataFrame({'ID': [1, 2, 3],'Name': ['Ram', 'Sam', 'John']})
df2 = pd.DataFrame({'ID': [1, 2, 3],'Marks': [80, 90, 85]})
print(df1)
print(df2)
df3 = pd.merge(df1, df2, on='ID')
print(df3)
```

8.

```python
data = {'Marks': [80, 90, 75, 85],'Age': [20, 21, 19, 22]}
index = [['A', 'A', 'B', 'B'], [1, 2, 1, 2]]
df = pd.DataFrame(data, index=index)
df.index.names = ['Group', 'ID']
print(df)
print(df.loc['A'])
print(df.loc['B', 1])
```

9.

```python
df = pd.DataFrame({
    'Category': ['A', 'A', 'B', 'B', 'A'],
    'SubCategory': ['X', 'Y', 'X', 'Y', 'X'],
    'Value': np.random.randint(1, 100, 5)
})
df = df.set_index(['Category', 'SubCategory'])
print(df)
print(df.groupby(level=['Category', 'SubCategory']).sum())
```

```python
df = pd.DataFrame(np.random.randint(1, 10, (5, 3)))
print(df)
df = df.apply(lambda x: x * 2)
print(df)
```

11.

```python
df = pd.DataFrame(np.random.randint(1, 10, (6, 3)),columns=['A', 'B', 'C'])
print(df)
df['Sum'] = df.apply(lambda x: x['A'] + x['B'] + x['C'], axis=1)
print(df)
```
