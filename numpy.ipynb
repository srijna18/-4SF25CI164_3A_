```python
import numpy as np
```

1. Array Creation and Manipulation

```python
a = np.random.randint(1, 21, (5, 5))
print(a)
a[:, 2] = 1
print(a)
```

2.

```python
a = np.arange(1, 17).reshape(4, 4)
print(a)
np.fill_diagonal(a, 0)
print(a)
```

3.

```python
a = np.random.randint(1, 21, (5, 5))
print(a)

print(a[0, :])
print(a[-1, :])
print(a[1:-1, 0])
print(a[1:-1, -1])
```

4.

```python
a = np.arange(1, 17).reshape(4, 4)
print(a)
print("Row sum:", np.sum(a, axis=1))
print("Column sum:", np.sum(a, axis=0))
```

5.

```python
a = np.arange(1, 10).reshape(3, 3)
print(a)
b = a.reshape(1, 9)
print(b)
c = a.reshape(9, 1)
print(c)
```

```python

a = np.array([
    ('Tom', 25, 65.5),
    ('Sam', 20, 60.2),
    ('Ram', 22, 70.5)
], dtype=[('name', 'U10'), ('age', 'i4'), ('weight', 'f4')])
print(a)
a.sort(order='age')
print(a)
```
